```python
#Data Selecting 
students = pd.DataFrame(
    {
        "name" : ["Ayushi","Manav","Jyoti","Tamana","Himanshu","Raj"],
        "branch" : ["SDS","SE","CSE","CES","DA","FSD"],
        "marks" : [92,98,45,66,86,np.nan],
        "hours_studied" : [8,9,6,7,5,4],
        "age" : [21,34,54,13,12,34],
    }
)
print("Marks of students\n",students['marks']) #column name - series
print()
students[['name','branch','marks']]
```

```python
students.loc[0:2,"name"] #students.loc[rows,cols] - inclusive of the end label
```

```python
students.iloc[0:3,0:2] #students.iloc[rows , cols] - end excluded
```

```python
students[students["marks"]<60] #boolean mask
```

```python
students[students["branch"] == "SDS"]
```

```python
students[(students["age"]>30) &(students["marks"]>60)] #AND operator
```

```python
display(students[(students["age"]>30) | (students["marks"]>60)]) #OR operator
```

```python
df = students.copy() #work on a copy , keep the original dataset

df["marks_percent"] = df["marks"]/100
print(df)
print()
print(df["marks_percent"])
```

```python
df["pass"] = df["marks"] >= 70
print(df)
```

```python
df['upper_name'] = df['name'].str.upper()
print(df['upper_name'])
print(df)
```

```python
df['lower_name'] = df['name'].str.lower()
print(df['lower_name'])
```

```python
df.sort_values("marks",ascending=False)  #Decending order 
```

```python
df.sort_values("marks",ascending=True) #Pure ascending order
```

```python
df["grade"] = pd.cut(df["marks"],
                     bins=[0,30,50,70,100],
                     labels=["fail","D","B","A"])
print(df[["marks","grade"]])
```

```python
#pandas has three basic moves : detect(isna) , drop(dropna) , fill(fillna)
print("missing per columns :\n",df.isna().sum())
```

```python
print("Rows with any missing :",df.isna().any(axis=1).sum()) #axis = 0 -> columns , axis = 1 -> rows
```

```python
print("columns with any missing :",df.isna().any(axis=0).sum()) 
```

```python
filled = df["marks"].fillna(df["marks"].median()) #median 
print(filled.tolist())
filled = df["marks"].fillna(df["marks"].mean()) #mean
print(filled.tolist())
filled = df["marks"].fillna(df["marks"].mode()) #mode
print(filled.tolist())
print(df)
```

```python
df.dropna() #removes all rows in dataset
df = df.drop(columns="passed")
print(df)
```

```python
df = df.drop(columns="upper_name")
print(df)
```

```python
df = df.drop(columns="lower_name")
print(df)
```
