Data Cleaning

```python

# ==========================================
# STEP 3: DATA CLEANING
# ==========================================

# 1. Check missing values in each column
df.isnull().sum()

# 2. Calculate percentage of missing values
(df.isnull().sum() / len(df)) * 100

# 3. Check total missing values in dataset
df.isnull().sum().sum()

# 4. Handle missing values in categorical columns
df["director"] = df["director"].fillna("Unknown")
df["cast"] = df["cast"].fillna("Unknown")
df["country"] = df["country"].fillna("Unknown")
df["rating"] = df["rating"].fillna("Not Rated")

# 5. Handle missing values in duration column
df["duration"] = df["duration"].fillna("Unknown")

# 6. Check duplicate records
df.duplicated().sum()

# 7. Remove duplicate records (if any)
df.drop_duplicates(inplace=True)

# 8. Convert date_added column into datetime format
df["date_added"] = pd.to_datetime(
    df["date_added"].str.strip(),
    errors="coerce"
)

# 9. Check updated data types
df.dtypes

# 10. Check missing values after cleaning
df.isnull().sum()

# 11. Check dataset information after cleaning
df.info()

# 12. Check final dataset dimensions
df.shape
```

Exploratory Data Analysis (EDA)

```python

# ==========================================
# STEP 4: EXPLORATORY DATA ANALYSIS
# ==========================================

# 1. Count Movies and TV Shows
df["type"].value_counts()

# 2. Calculate percentage of Movies and TV Shows
df["type"].value_counts(normalize=True) * 100

# 3. Find top 10 release years
df["release_year"].value_counts().head(10)

# 4. Count content released in each year
df["release_year"].value_counts().sort_index()

# 5. Find top 10 countries
country_counts = (
    df["country"]
    .replace("Unknown", np.nan)
    .dropna()
    .str.split(", ")
    .explode()
    .value_counts()
)
country_counts.head(10)

# 6. Find top 10 genres
genre_counts = (
    df["listed_in"]
    .str.split(", ")
    .explode()
    .value_counts()
)
genre_counts.head(10)

# 7. Find top 10 directors
# Find top 10 directors after excluding missing/Unknown values
df.loc[df["director"] != "Unknown", "director"].value_counts().head(10)

# 8. Analyze content ratings
df["rating"].value_counts()

# 9. Analyze content added in each year
df["year_added"] = df["date_added"].dt.year
df["year_added"].value_counts().sort_index()

# 10. Compare Movies and TV Shows added each year
df.groupby(["year_added", "type"]).size()

# 11. Calculate average movie duration
movies = df[df["type"] == "Movie"].copy()

movies["duration_minutes"] = (
    movies["duration"]
    .str.extract(r"(\d+)")
    .astype(float)
)
movies["duration_minutes"].mean()

# 12. Calculate average TV Show seasons
tv = df[df["type"] == "TV Show"].copy()

tv["seasons"] = (
    tv["duration"]
    .str.extract(r"(\d+)")
    .astype(float)
)
tv["seasons"].mean()

# 13. Find the oldest and newest release years
df["release_year"].min()
df["release_year"].max()

# 14. Find the most common content rating
df["rating"].mode()
```

Data Visualization

```python

# ==========================================
# STEP 5: DATA VISUALIZATION
# ==========================================

# 1. Movies vs TV Shows - Count Plot
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="type")
plt.title("Movies vs TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
plt.show()


# 2. Top 10 Countries - Bar Chart
plt.figure(figsize=(10, 5))
country_counts.head(10).plot(kind="bar")
plt.title("Top 10 Countries by Netflix Titles")
plt.xlabel("Country")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()


# 3. Top 10 Genres - Horizontal Bar Chart
plt.figure(figsize=(10, 5))
genre_counts.head(10).plot(kind="barh")
plt.title("Top 10 Netflix Genres")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")
plt.gca().invert_yaxis()
plt.show()


# 4. Content Added Over the Years - Line Chart
plt.figure(figsize=(10, 5))
yearly_content = df["year_added"].value_counts().sort_index()
yearly_content.plot(kind="line", marker="o")
plt.title("Netflix Content Added Over the Years")
plt.xlabel("Year")
plt.ylabel("Number of Titles")
plt.grid(True)
plt.show()


# 5. Movies vs TV Shows - Pie Chart
plt.figure(figsize=(7, 7))
df["type"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)
plt.title("Netflix Content Distribution")
plt.ylabel("")
plt.show()


# 6. Content Ratings - Bar Chart
plt.figure(figsize=(10, 5))
df["rating"].value_counts().plot(kind="bar")
plt.title("Netflix Content Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()


# 7. Movie Duration Distribution - Histogram
plt.figure(figsize=(10, 5))
sns.histplot(
    data=movies,
    x="duration_minutes",
    bins=30,
    kde=True
)
plt.title("Distribution of Movie Durations")
plt.xlabel("Duration (Minutes)")
plt.ylabel("Frequency")
plt.show()
```
